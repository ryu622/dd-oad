## 4. 結果の確認

```python
import json

with open('../documents/pooled_real_result.json') as f:
    result = json.load(f)

print(f"n_events = {result['n_events']}")
print(f"f_d0    = {result['f_d0']:.3f}")
print(f"f_d_inf = {result['f_d_inf']:.3f}")
print(f"d0      = {result['d0']:.3f}")
print(f"mean fitting error = {result['mean_error']:.4f}")
print(f"elapsed = {result['elapsed_min']:.1f} min")
print()
if result['f_d0'] > result['f_d_inf']:
    print("-> H1' supported (近いほど f_d が大きい)")
else:
    print("-> H1' と逆方向 (遠いほど f_d が大きい)")
```
